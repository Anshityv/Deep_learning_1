## Part 1: Theory Questions

### 1. What is deep learning, and how is it connected to artificial intelligence?
* **Deep Learning:** A specialized subset of machine learning based on artificial neural networks with multiple layers (hence "deep") that attempt to model high-level abstractions in data. It automatically extracts features from raw data without requiring manual human feature engineering.
* **Connection to AI:** Artificial Intelligence (AI) is the broad overarching field of creating systems capable of mimicking human intelligence. Machine learning is a subset of AI that enables systems to learn from data, and deep learning is a sub-field of machine learning that powers state-of-the-art AI applications like computer vision, natural language processing, and generative AI.

### 2. What is a neural network, and what are the different types of neural networks?
* **Neural Network:** An artificial neural network (ANN) is a computational model inspired by the human brain. It consists of interconnected nodes (neurons) organized into layers: an input layer, one or more hidden layers, and an output layer. Each connection has a weight that adjusts during training.
* **Different Types of Neural Networks:**
  * **Feedforward Neural Network (FNN):** Information moves in only one direction—forward from the input, through the hidden layers, to the output.
  * **Convolutional Neural Network (CNN):** Specialized for processing grid-like data such as images, using convolutional layers to capture spatial patterns.
  * **Recurrent Neural Network (RNN):** Designed for sequential data (like time series or text), featuring loops that allow information to persist.
  * **Long Short-Term Memory (LSTM):** An advanced variant of RNNs capable of learning long-term dependencies.
  * **Transformer:** A modern architecture relying entirely on attention mechanisms, dominant in modern Large Language Models.

### 3. What is the mathematical structure of a neural network?
Mathematically, a single artificial neuron computes a weighted sum of its inputs, adds a bias term, and passes the result through a non-linear activation function. For a given neuron:

$$z = \sum_{i=1}^{n} (w_i x_i) + b$$

$$a = f(z)$$

Where:
* $x_i$ represents the input features.
* $w_i$ represents the corresponding weights.
* $b$ is the bias term.
* $z$ is the linear combination.
* $f$ is the activation function.
* $a$ is the final output of the neuron.

Across a network, these operations are represented compactly using matrix multiplications and vector additions across multiple layers.

### 4. What is an activation function, and why is it essential in neural network?
* **Activation Function:** A mathematical function applied to the output of a neuron's linear transformation before passing it to the next layer.
* **Why it is Essential:** Without activation functions, a neural network—regardless of how many layers it has—would behave just like a single-layer linear regression model. Activation functions introduce non-linearity into the network, enabling it to learn complex, non-linear patterns, decision boundaries, and relationships in the data.

### 5. Could you list some common activation functions used in neural networks?
* **Sigmoid:** Outputs values between $0$ and $1$, often used for binary classification output layers. Formula: $\sigma(x) = \frac{1}{1 + e^{-x}}$
* **ReLU (Rectified Linear Unit):** Outputs $x$ if $x > 0$ and $0$ otherwise. It helps mitigate the vanishing gradient problem. Formula: $f(x) = \max(0, x)$
* **Leaky ReLU:** A variant of ReLU that allows a small, non-zero gradient when the input is negative.
* **Tanh (Hyperbolic Tangent):** Outputs values between $-1$ and $1$. Formula: $\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$
* **Softmax:** Used in multi-class classification output layers to convert raw scores into probability distributions that sum to $1$.

### 6. What is a multilayer neural network?
A multilayer neural network (often called a Multilayer Perceptron or MLP) is a neural network featuring one or more hidden layers positioned between the input layer and the output layer. The presence of hidden layers allows the network to learn hierarchical feature representations, where earlier layers capture simple patterns (like edges or basic traits) and deeper layers combine them to form complex concepts.

### 7. What is a loss function, and why is it crucial for neural network training?
* **Loss Function:** A mathematical metric used to measure the difference or error between the model's predicted output and the actual ground truth labels.
* **Why it is Crucial:** The loss function provides quantitative feedback on how poorly or well the network is performing. This scalar error value is what optimization algorithms (like gradient descent) aim to minimize by iteratively updating the network's weights and biases.

### 8. What are some common types of loss functions?
* **For Regression Tasks:**
  * **Mean Squared Error (MSE):** Measures the average squared difference between estimated values and actual value.
  * **Mean Absolute Error (MAE):** Measures the average absolute magnitude of the errors.
* **For Classification Tasks:**
  * **Binary Cross-Entropy:** Used for binary classification problems ($2$ classes).
  * **Categorical Cross-Entropy / Sparse Categorical Cross-Entropy:** Used for multi-class classification problems (> $2$ classes).

### 9. How does a neural network learn?
A neural network learns through an iterative optimization loop consisting of three core steps:
1. **Forward Propagation:** Input data is passed through the network layers to generate a prediction.
2. **Loss Calculation:** The error between the prediction and the true label is computed using a loss function.
3. **Backward Propagation and Optimization:** Gradients of the loss with respect to each weight are calculated using the chain rule (backpropagation), and an optimizer updates the weights to reduce the loss in subsequent iterations.

### 10. What is an optimizer in neural networks, and why is it necessary?
* **Optimizer:** An optimization algorithm or method used to update the weights and biases of a neural network by minimizing the loss function.
* **Why it is Necessary:** While the loss function tells the model how wrong it is, the optimizer determines how to change the parameters (weights and biases) efficiently to correct mistakes, navigating the high-dimensional loss landscape to find a global or local minimum.

### 11. Could you briefly describe some common optimizers?
* **Gradient Descent (GD):** Computes the gradient of the loss function for the entire dataset to update weights. Can be slow for large datasets.
* **Stochastic Gradient Descent (SGD):** Updates parameters using only a single sample or a small batch at a time, making it faster and able to escape local minima due to noise.
* **RMSprop (Root Mean Square Propagation):** Adapts the learning rate for each parameter by scaling it based on a running average of recent gradient magnitudes.
* **Adam (Adaptive Moment Estimation):** Combines the benefits of both RMSprop and AdaGrad by keeping tracks of exponential moving averages of past gradients (momentum) and squared gradients. It is widely considered the default standard optimizer.

### 12. Can you explain forward and backward propagation in a neural network?
* **Forward Propagation:** The process where input data flows forward through the network from the input layer, passes through hidden layers (undergoing weighted sums and activation functions), and produces an output prediction at the final layer.
* **Backward Propagation (Backpropagation):** The reverse process used during training. It computes the derivative (gradient) of the loss function with respect to each weight and bias by applying the chain rule backward from the output layer to the input layer. These gradients tell the optimizer the direction and magnitude to adjust weights.

### 13. What is weight initialization, and how does it impact training?
* **Weight Initialization:** The method or statistical distribution used to set the initial random values of a neural network's weights before training begins.
* **Impact on Training:** Poor weight initialization can cause signals to amplify exponentially (exploding gradients) or shrink to zero (vanishing gradients), rendering training extremely slow or entirely stagnant. Good initialization ensures signals flow properly through deep layers right from the start.

### 14. What is the vanishing gradient problem in deep learning?
The vanishing gradient problem occurs in deep neural networks during backpropagation when gradients become exponentially small as they flow backward from the output layer to the earlier layers. Because the gradients approach zero, the weights of the initial layers barely update, effectively halting their learning and preventing the network from training successfully. This often happens with saturating activation functions like Sigmoid or Tanh.

### 15. What is the exploding gradient problem?
The exploding gradient problem occurs when large error gradients accumulate during backpropagation, causing exponential growth in weight updates from one iteration to the next. This results in massive updates to model parameters, making the network unstable, causing loss values to oscillate wildly, or resulting in numerical overflow (producing NaN values).

## Part 2: Practical Questions

### 1. How do you create a simple perceptron for basic binary classification?

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

# Creating a single perceptron model
input_dim = 10 # Example dimension
model = Sequential([
    Dense(units=1, input_shape=(input_dim,), activation='sigmoid')
])

model.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy'])
```

### 2. How can you build a neural network with one hidden layer using Keras?

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

input_dim = 10 # Example dimension
model = Sequential([
    Dense(units=64, activation='relu', input_shape=(input_dim,)), # Hidden layer
    Dense(units=1, activation='sigmoid')                          # Output layer
])
```

### 3. How do you initialize weights using the Xavier (Glorot) initialization method in Keras?

```python
from tensorflow.keras.layers import Dense

# Adding layer with Xavier initialization
model.add(Dense(units=64, activation='relu', kernel_initializer='glorot_uniform'))
```

### 4. How can you apply different activation functions in a neural network in Keras?

```python
from tensorflow.keras.layers import Dense, Activation

# Method 1: Inside the layer definition
model.add(Dense(64, activation='relu'))

# Method 2: As a separate layer
model.add(Dense(64))
model.add(Activation('relu'))
```

### 5. How do you add dropout to a neural network model to prevent overfitting?

```python
from tensorflow.keras.layers import Dropout

model.add(Dense(64, activation='relu'))
model.add(Dropout(0.5)) # Drops 50% of the neurons randomly during training
```

### 6. How do you manually implement forward propagation in a simple neural network?

```python
import numpy as np

# Sample data and parameters
np.random.seed(42)
X = np.array([[0.5, 1.2]])
W1 = np.random.randn(2, 3) # Weights for hidden layer
b1 = np.zeros((1, 3))      # Bias for hidden layer
W2 = np.random.randn(3, 1) # Weights for output layer
b2 = np.zeros((1, 1))      # Bias for output layer

# Activation function (ReLU)
def relu(x):
    return np.maximum(0, x)

# Forward pass
z1 = np.dot(X, W1) + b1
a1 = relu(z1)
z2 = np.dot(a1, W2) + b2
output = 1 / (1 + np.exp(-z2)) # Sigmoid output
print("Forward propagation output:", output)
```

### 7. How do you add batch normalization to a neural network model in Keras?

```python
from tensorflow.keras.layers import Dense, BatchNormalization, Activation

model.add(Dense(64))
model.add(BatchNormalization())
model.add(Activation('relu'))
```

### 9. How can you use gradient clipping in Keras to control the gradient size and prevent exploding gradients?

```python
from tensorflow.keras.optimizers import Adam

optimizer = Adam(learning_rate=0.001, clipvalue=1.0) # Clips gradients between -1.0 and 1.0
model.compile(optimizer=optimizer, loss='binary_crossentropy')
```

### 10. How can you create a custom loss function in Keras?

```python
import tensorflow as tf

def custom_mean_squared_error(y_true, y_pred):
    return tf.reduce_mean(tf.square(y_true - y_pred))

model.compile(optimizer='adam', loss=custom_mean_squared_error)
```
